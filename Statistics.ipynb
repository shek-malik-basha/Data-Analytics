```
import pandas as pd
df = pd.read_csv('Walmart_Sales.csv')

range_sales = df['Weekly_Sales'].max() - df['Weekly_Sales'].min()
print("Range:", range_sales)
```

```

varience = df['Weekly_Sales'].var()
print("Varience:", varience)
```

```
std = df['Weekly_Sales'].std()
print("Standard Deviation:", std)
```

```
mean = df['Weekly_Sales'].mean()
print("Mean:",mean)
```

```
median = df['Weekly_Sales'].median()
print("Median:", median)
```

```
mode = df['Weekly_Sales'].mode()
print("Mode:", mode)
```

```
mean = df['Weekly_Sales'].mean()

mad = (df['Weekly_Sales'] - mean).abs().mean()

print(mad)
```

```
df['Weekly_Sales'].quantile([0.25,0.5,0.75])
```

```
Q1 = df['Weekly_Sales'].quantile(0.25)
Q3 = df['Weekly_Sales'].quantile(0.75)

IQR = Q3 - Q1

print(IQR,"is IQR")
```
