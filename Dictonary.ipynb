```python
input_dict = {"name":"monthy","profession":"singer"}
key = input("enter key")
if key in input_dict.keys():
  print(input_dict.values())
else:
  print("NA")
```

```python
input_dict = {"name":"monthy","profession":"singer"}
key = input("enter key")
if key in input_dict.keys():
  print(input_dict[key])
else:
  print("NA")
```
