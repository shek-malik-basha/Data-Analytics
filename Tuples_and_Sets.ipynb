**Sorting**

```python
#sorting
t6 = (1,2,8,90,34)
t_new = sorted(t6)
print("sorted:",t_new)
print(type(t_new))
```

**Sets**
Set is a built-in Python data type used to store a collection of unique items.

->Stores only unique elements; duplicate values are automatically removed.
->Unordered collection, so elements do not have a fixed position and cannot be accessed using indexes.
->Supports fast search, insertion and deletion operations using hashing internally.


```python
A = {5,1,3,4,4,5,6,7}
B = {3,3,5,5,1,7,2}
C = A & B
print(C)
print(sorted(C))
```

```python
A = [5,1,3,4,4,5,6,7]
B = [3,3,5,5,1,7,2]
C = set(A) & set(B)
print(sorted(C))
```

```python
nums = set([1,1,2,3,3,3,4])
print(len(nums))
```

```python
s = set(["a","b","c","d"])
print(s)
```

```python
a = {"x", "y"}
b = {"y", "z"}
u = a.union(b)
print(u)
```

```python
a = {1, 2, 3}
b = {2, 3, 4}
i = a.intersection(b)
print(i)
```

```python
a = {1, 2, 3}
b = {2, 3, 4}
d = a.difference(b)
print(d)
```

```python
s = {1, 2, 3}
s.discard(2)
print(s)
```

```python
a = {1,2,3}
b = {4,5,6}
print(a.isdisjoint(b))
```

```python
a = {1,2,3,4}
b= {1,2,3,4,5,6,7}
print(a.issubset(b))
```
